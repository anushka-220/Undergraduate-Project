```python
import numpy as np
import pandas as pd
from sklearn.neighbors import NearestNeighbors

def compute_ligand_receptor_matrix(adata, ligands, receptors, k=10):
    # Step 1: Get spatial coordinates
    coords = adata.obsm['spatial'][:, :2]

    # Step 2: Find k nearest neighbors (excluding self)
    nbrs = NearestNeighbors(n_neighbors=k+1, algorithm='kd_tree').fit(coords)
    _, indices = nbrs.kneighbors(coords)
    neighbor_indices = indices[:, 1:]  # remove self
    adata_upper = adata.copy()
    adata_upper.var_names = [gene.upper() for gene in adata.var_names]

    X_lig = adata_upper[:, ligands].X.toarray()
    X_rec = adata_upper[:, receptors].X.toarray()


    # Step 4: Initialize matrices
    n_cells = adata.n_obs
    ligand_matrix = np.zeros((n_cells, len(ligands)))
    receptor_matrix = np.zeros((n_cells, len(receptors)))

    # Step 5: For each cell, sum expressions from its neighbors
    for i in range(n_cells):
        neighbors = neighbor_indices[i]
        ligand_matrix[i, :] = X_lig[neighbors].sum(axis=0)
        receptor_matrix[i, :] = X_rec[neighbors].sum(axis=0)

    # Step 6: Combine into one big matrix
    combined_matrix = np.hstack([ligand_matrix, receptor_matrix])
    columns = [f"g_ligand_{l}" for l in ligands] + [f"g_receptor_{r}" for r in receptors]
    df = pd.DataFrame(combined_matrix, columns=columns, index=adata.obs_names)

    return df
```

```python

df = compute_ligand_receptor_matrix(adata, ligand_expr, receptor_expr, k=10)
```
