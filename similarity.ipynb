```python

def preprocess_data(adata: AnnData, n_top_genes: int = 2048):
    raw_data = AnnData(adata.X.copy())
    raw_data.obs_names, raw_data.var_names = adata.obs_names, adata.var_names
    adata.raw = raw_data

    sc.pp.normalize_per_cell(adata)
    sc.pp.log1p(adata)
    sc.pp.highly_variable_genes(adata, n_top_genes=n_top_genes)
```

```python
def calculate_neighbors(adata: AnnData, n_neighbors: int = 15):
    sc.pp.neighbors(adata, n_neighbors=n_neighbors)
    sc.tl.umap(adata)
    return adata
```

```python
def flatten_covet(adata: AnnData):
    if 'COVET_SQRT' in adata.obsm:
        flattened_array = adata.obsm['COVET_SQRT'].reshape(adata.obsm['COVET_SQRT'].shape[0], -1)
        adata.obsm['COVET_SQRT_Flattened'] = flattened_array
    else:
        raise KeyError("COVET_SQRT not found in adata.obsm")
    return adata
```
